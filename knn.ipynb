```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
%matplotlib inline
```

```python
from sklearn.datasets import load_iris
```

```python
df = load_iris()
```

```python
df1 = pd.DataFrame(df , columns=df.target)
```

```python
x = df.data
y = df.target
```

```python
from sklearn.model_selection import train_test_split
```

```python
x_train , x_test , y_train , y_test = train_test_split(x,y, test_size= 0.25 , random_state=42)
```

```python
from sklearn.neighbors import KNeighborsClassifier

knn = KNeighborsClassifier(n_neighbors=5)
```

```python
from sklearn.preprocessing import StandardScaler

ss = StandardScaler()
```

```python
ss.fit_transform(x_train)
ss.fit(x_test)
```

```python
from sklearn.model_selection import GridSearchCV

hparam = {"n_neighbors":[3,5,7,11]}

regressor = GridSearchCV(knn , param_grid= hparam ,cv = 5)

regressor.fit(x_train , y_train)
```

```python
predict_Class = regressor.predict(x_test)
```

```python
from sklearn.metrics import accuracy_score , classification_report

print("Accuracy : ", accuracy_score(y_test , predict_Class))
print("Classification report :" , classification_report(y_test , predict_Class))
```

```python
print(regressor.best_score_)
print(regressor.best_params_)
```
